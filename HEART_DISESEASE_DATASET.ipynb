```python
# Import necessary libraries
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from fpdf import FPDF
import kagglehub

# Download the latest version of the dataset
path = kagglehub.dataset_download("johnsmith88/heart-disease-dataset")
print("Path to dataset files:", path)

# Load the dataset
heart_data = pd.read_csv(path + '/heart.csv')

# Visualizations
# Set the style for seaborn
sns.set(style="whitegrid")

# Create a PDF document to compile all visualizations
pdf = FPDF()
pdf.set_auto_page_break(auto=True, margin=15)
pdf.add_page()
pdf.set_font('Arial', 'B', 16)
pdf.cell(0, 10, 'Heart Disease Analysis Visualizations', ln=True, align='C')

# Maximum Heart Rate by Heart Disease Presence
plt.figure(figsize=(10, 6))
sns.boxplot(x='target', y='thalach', data=heart_data)
plt.title('Maximum Heart Rate by Heart Disease Presence')
plt.xlabel('Heart Disease (0 = No, 1 = Yes)')
plt.ylabel('Maximum Heart Rate')
plt.savefig('maximum_heart_rate.png')
plt.close()
pdf.cell(0, 10, 'Maximum Heart Rate by Heart Disease Presence', ln=True)
pdf.image('maximum_heart_rate.png', x=10, w=190)

# Exercise Angina by Heart Disease Presence
plt.figure(figsize=(10, 6))
sns.countplot(x='exang', hue='target', data=heart_data)
plt.title('Exercise Angina by Heart Disease Presence')
plt.xlabel('Exercise Angina (0 = No, 1 = Yes)')
plt.ylabel('Count')
plt.legend(title='Heart Disease (0 = No, 1 = Yes)')
plt.savefig('exercise_angina.png')
plt.close()
pdf.cell(0, 10, 'Exercise Angina by Heart Disease Presence', ln=True)
pdf.image('exercise_angina.png', x=10, w=190)

# Cholesterol Levels by Heart Disease Presence
plt.figure(figsize=(10, 6))
sns.boxplot(x='target', y='chol', data=heart_data)
plt.title('Cholesterol Levels by Heart Disease Presence')
plt.xlabel('Heart Disease (0 = No, 1 = Yes)')
plt.ylabel('Cholesterol Level')
plt.savefig('cholesterol_levels.png')
plt.close()
pdf.cell(0, 10, 'Cholesterol Levels by Heart Disease Presence', ln=True)
pdf.image('cholesterol_levels.png', x=10, w=190)

# Chest Pain Types by Heart Disease Presence
plt.figure(figsize=(10, 6))
sns.countplot(x='cp', hue='target', data=heart_data)
plt.title('Chest Pain Types by Heart Disease Presence')
plt.xlabel('Chest Pain Type')
plt.ylabel('Count')
plt.legend(title='Heart Disease (0 = No, 1 = Yes)')
plt.savefig('chest_pain_types.png')
plt.close()
pdf.cell(0, 10, 'Chest Pain Types by Heart Disease Presence', ln=True)
pdf.image('chest_pain_types.png', x=10, w=190)

# Age Distribution by Heart Disease Presence
plt.figure(figsize=(10, 6))
sns.boxplot(x='target', y='age', data=heart_data)
plt.title('Age Distribution by Heart Disease Presence')
plt.xlabel('Heart Disease (0 = No, 1 = Yes)')
plt.ylabel('Age')
plt.savefig('age_distribution.png')
plt.close()
pdf.cell(0, 10, 'Age Distribution by Heart Disease Presence', ln=True)
pdf.image('age_distribution.png', x=10, w=190)

# Save the PDF
pdf_file_path = 'heart_disease_visualizations.pdf'
pdf.output(pdf_file_path)
print('PDF created successfully with all visualizations.')
```
